```python
import pandas as pd, numpy as np
import vivarium_inputs
import gbd_mapping
import pathlib
```

```python
location = "India"
vehicle = "rice"
```

```python
# Parameters
location = "nigeria"
vehicle = "bouillon"
```

```python
location = location.title()
```

```python
pop = vivarium_inputs.get_population_structure(location).value
pop[pop > 0]
```

```python
asfr = vivarium_inputs.get_measure(gbd_mapping.covariates.age_specific_fertility_rate, "estimate", location).value
asfr[asfr > 0]
```

```python
asfr = asfr[asfr.index.get_level_values("parameter") == "mean_value"].droplevel("parameter")
asfr
```

```python
sbr = vivarium_inputs.get_measure(gbd_mapping.covariates.stillbirth_to_live_birth_ratio, "estimate", location).value
sbr[sbr > 0]
```

```python
sbr = sbr.iloc[0]
sbr
```

```python
maternal_abortion_miscarriage_incidence = vivarium_inputs.get_measure(
    gbd_mapping.causes.maternal_abortion_and_miscarriage,
    "incidence_rate",
    location,
).mean(axis=1)
maternal_abortion_miscarriage_incidence[maternal_abortion_miscarriage_incidence > 0]
```

```python
ectopic_pregnancy_incidence = vivarium_inputs.get_measure(
    gbd_mapping.causes.ectopic_pregnancy,
    "incidence_rate",
    location,
).mean(axis=1)
ectopic_pregnancy_incidence[ectopic_pregnancy_incidence > 0]
```

```python
pregnancy_incidence = (
    asfr
    + (asfr * sbr)
    + maternal_abortion_miscarriage_incidence
    + ectopic_pregnancy_incidence
)
pregnancy_incidence[pregnancy_incidence > 0]
```

```python
pregnancies = (pop * pregnancy_incidence)
pregnancies[pregnancies > 0]
```

```python
total_pregnancies = pregnancies.sum()
total_pregnancies
```

```python
sim_population = pd.read_parquet(f"../../0200_pregnancy_sim/sim_results/{vehicle}/{location.lower()}/pregnancy_outcome_count.parquet").groupby("input_draw").value.sum().mean()
sim_population
```

```python
scalar = total_pregnancies / sim_population
scalar
```

```python
for result in ["ylds", "ylls", "pregnancy_outcome_count"]:
    df = pd.read_parquet(f"../../0200_pregnancy_sim/sim_results/{vehicle}/{location.lower()}/{result}.parquet")
    df.value *= scalar
    path = pathlib.Path(f'./pregnancy_results/{vehicle}/{location.lower()}/{result}.parquet')
    path.parent.mkdir(exist_ok=True, parents=True)
    df.to_parquet(path)
```
