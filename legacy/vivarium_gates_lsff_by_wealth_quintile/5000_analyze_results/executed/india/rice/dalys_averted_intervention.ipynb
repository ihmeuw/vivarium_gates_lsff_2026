```python
def aggregate_by_scenario(df):
    return df.groupby(["scenario", "input_draw", "wealth_quintile","sub_entity"]).value.sum().groupby(["scenario", "wealth_quintile","sub_entity"]).mean()
```

```python
# Calculate total person-time for pregnant people
anemia_person_time = pd.read_parquet(f"0100_rescale_results/pregnancy_results/{vehicle}/{location}/person_time_anemia.parquet")
anemia_person_time = aggregate_by_scenario(anemia_person_time)
total_preg_person_time = anemia_person_time.groupby(level=['wealth_quintile']).sum()
total_preg_person_time
```

```python
# Calculate total person-time for children
child_person_time = pd.read_parquet(f"0100_rescale_results/child_results/{vehicle}/{location}/person_time.parquet")
child_person_time = child_person_time.groupby(["child_scenario", "input_draw", "wealth_quintile"]).value.sum().groupby(["child_scenario", "wealth_quintile"]).mean()
total_child_person_time = child_person_time.groupby(level=['wealth_quintile']).sum()
total_child_person_time
```

```python
# Calculate total person-time for both pregnant people and children
total_person_time = (total_child_person_time + total_preg_person_time).groupby(["wealth_quintile"]).mean()
total_person_time
```

```python
df_dalys = pd.DataFrame(dalys_averted).rename(columns={'value':'total_dalys_averted'})
df_dalys
```

```python
df_dalys['total_dalys_averted_rate'] = df_dalys.total_dalys_averted / total_person_time
df_dalys['pregnancy_dalys_averted_rate'] = pregnancy_dalys_averted / total_preg_person_time
df_dalys['child_dalys_averted_rate'] = child_dalys_averted / total_child_person_time
```

```python
df_dalys['pregnancy_dalys_averted'] = pregnancy_dalys_averted
df_dalys['child_dalys_averted'] = child_dalys_averted
df_dalys
```

```python
df_dalys['pct_child_dalys_averted'] = (child_dalys_averted / child_ylls_by_scenario.loc["baseline"]) * 100
df_dalys
```

```python
df_dalys['pct_pregnancy_dalys_averted'] = (pregnancy_dalys_averted / baseline_pregnancy_dalys) * 100
df_dalys
```

```python
df_dalys['pct_total_dalys_averted'] = (df_dalys.total_dalys_averted / baseline_dalys) * 100
df_dalys
```

```python
order = ['lowest', 'second', 'middle', 'fourth', 'highest']
df_dalys = df_dalys.reindex(order)
df_dalys
```
