```python
import pandas as pd, numpy as np
```

```python
df = pd.DataFrame({
    'scaled_surv_0': [0.5, 0.75],
    'scaled_surv_1': [0.4, 0.75],
    'lambda_years': [1, 2],
})
df
```

```python
num_draws = 2
```

```python
df[[f'surv_abs_{i}' for i in range(num_draws)]] = df.filter(like='scaled_surv_').mul(np.exp(df.lambda_years), axis=0).rename(columns=lambda c: c.replace('scaled_surv_', 'surv_abs_'))
df
```
