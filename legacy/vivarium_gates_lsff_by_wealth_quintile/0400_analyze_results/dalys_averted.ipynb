```python
import pandas as pd
```

```python
# NOTE: **raw** DALYs, not yet scaled to population size!
for location in ["india", "nigeria"]:
    pregnancy_ylls_by_scenario = pd.read_parquet(f'../0200_pregnancy_sim/sim_results/{location}/ylls.parquet').groupby('scenario').value.sum()
    print(f'{location.title()} pregnancy YLLs')
    display(pregnancy_ylls_by_scenario.sort_values())
    pregnancy_ylds_by_scenario = pd.read_parquet(f'../0200_pregnancy_sim/sim_results/{location}/ylds.parquet').groupby('scenario').value.sum()
    print(f'{location.title()} pregnancy YLDs')
    display(pregnancy_ylds_by_scenario.sort_values())
    pregnancy_dalys_by_scenario = pregnancy_ylls_by_scenario + pregnancy_ylds_by_scenario
    print(f'{location.title()} pregnancy DALYs')
    display(pregnancy_dalys_by_scenario.sort_values())

    child_ylls_by_scenario = pd.read_parquet(f'../0300_child_sim/sim_results/{location}/ylls.parquet').rename(columns={'maternal_scenario': 'scenario'}).groupby('scenario').value.sum()
    print(f'{location.title()} child YLLs')
    display(child_ylls_by_scenario.sort_values())
    child_dalys_by_scenario = child_ylls_by_scenario
    print(f'{location.title()} child DALYs')
    display(child_dalys_by_scenario.sort_values())

    print(f'{location.title()} DALYs')
    display((pregnancy_dalys_by_scenario + child_dalys_by_scenario).sort_values())
```
