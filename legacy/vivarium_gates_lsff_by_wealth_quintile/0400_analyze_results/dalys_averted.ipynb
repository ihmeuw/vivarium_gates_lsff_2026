```python
import pandas as pd
```

```python
# NOTE: **raw** DALYs, not yet scaled to population size!
for location in ["india", "nigeria"]:
    pregnancy_dalys_by_scenario = pd.concat([
        pd.read_parquet(f'../0200_pregnancy_sim/sim_results/{location}/ylls.parquet'),
        pd.read_parquet(f'../0200_pregnancy_sim/sim_results/{location}/ylds.parquet'),
    ]).groupby('scenario').value.sum()
    child_dalys_by_scenario = pd.concat([
        pd.read_parquet(f'../0300_child_sim/sim_results/{location}/ylls.parquet').drop(columns=["child_scenario"]).rename(columns={"maternal_scenario": "scenario"}),
        pd.read_parquet(f'../0300_child_sim/sim_results/{location}/ylds.parquet').drop(columns=["child_scenario"]).rename(columns={"maternal_scenario": "scenario"}),
    ]).groupby('scenario').value.sum()
    print(f'{location.title()} DALYs')
    display((pregnancy_dalys_by_scenario + child_dalys_by_scenario).sort_values())
```
