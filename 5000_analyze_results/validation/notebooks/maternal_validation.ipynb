# Maternal (pregnancy) simulation — automated V&V

Baseline-scenario sim output vs. the maternal artifact, using only the built-in
`RatioMeasure` classes. See `../README.md` for the measure triage.

```python
import warnings

warnings.filterwarnings("ignore", category=FutureWarning)

from vivarium.validation import ValidationContext

from lsff_utils import paths
```

```python
LOCATION = "nigeria"
RUN = ""  # psimulate run timestamp; blank picks the most recent

location_root = paths.MATERNAL_RESULTS_ROOT / LOCATION
if RUN:
    RESULTS_DIR = location_root / RUN
else:
    RESULTS_DIR = max(
        (d for d in location_root.iterdir() if (d / "model_specification.yaml").exists()),
        key=lambda d: d.name,
    )

REPORT_PATH = f"maternal_validation_report_{LOCATION}.html"
print(RESULTS_DIR)
```

```python
# psimulate names branch columns after the config-path leaf: intervention.scenario -> 'scenario'.
vc = ValidationContext(results_dir=RESULTS_DIR, scenario_columns=["scenario"])
```

## Available simulation outputs

```python
vc.get_sim_outputs()
```

## Available artifact keys

```python
vc.get_artifact_keys()
```

## Comparisons — baseline, sim vs artifact

```python
BASELINE = {"scenario": "baseline"}

vc.add_comparison(
    "cause.maternal_disorders.cause_specific_mortality_rate",
    test_source="sim",
    ref_source="artifact",
    test_scenarios=BASELINE,
)

sorted(vc.comparisons)
```

## Report

```python
vc.generate_results(output_path=REPORT_PATH)
```
